```python
# STEP 4.6: EXPLICIT DATE PARSING AND CHRONOLOGY VALIDATION

date_columns = ["order_date", "dispatch_date", "delivery_date"]

# Re-parse from original strings, preserving alignment after duplicate removal
for col in date_columns:
    clean_df[col] = pd.to_datetime(
        df.loc[clean_df.index, col],
        format="%m/%d/%Y",
        errors="coerce"
    )

# Identify chronology issues
missing_order = clean_df["order_date"].isna()

dispatch_before_order = (
    clean_df["order_date"].notna()
    & (clean_df["dispatch_date"] < clean_df["order_date"])
)

delivery_before_dispatch = (
    clean_df["dispatch_date"].notna()
    & (clean_df["delivery_date"] < clean_df["dispatch_date"])
)

delivery_before_order = (
    clean_df["order_date"].notna()
    & (clean_df["delivery_date"] < clean_df["order_date"])
)

print("=" * 60)
print("DATE VALIDATION SUMMARY")
print("=" * 60)
print(f"Total cleaned shipments: {len(clean_df)}")
print(f"Missing order dates: {missing_order.sum()}")
print(f"Dispatch before order: {dispatch_before_order.sum()}")
print(f"Delivery before dispatch: {delivery_before_dispatch.sum()}")
print(f"Delivery before order: {delivery_before_order.sum()}")

print("\n--- Dispatch before order: sample records ---")
display(clean_df.loc[
    dispatch_before_order,
    ["shipment_id", "order_date", "dispatch_date", "delivery_date"]
].head(15))

print("\n--- Delivery before dispatch: sample records ---")
display(clean_df.loc[
    delivery_before_dispatch,
    ["shipment_id", "order_date", "dispatch_date", "delivery_date"]
].head(15))

print("\n--- Delivery before order: sample records ---")
display(clean_df.loc[
    delivery_before_order,
    ["shipment_id", "order_date", "dispatch_date", "delivery_date"]
].head(15))
```

```python
# STEP 4.7: CREATE DATA-QUALITY FLAGS

clean_df["flag_missing_order_date"] = clean_df["order_date"].isna()

clean_df["flag_dispatch_before_order"] = (
    clean_df["order_date"].notna()
    & (clean_df["dispatch_date"] < clean_df["order_date"])
)

clean_df["flag_delivery_before_dispatch"] = (
    clean_df["dispatch_date"].notna()
    & (clean_df["delivery_date"] < clean_df["dispatch_date"])
)

clean_df["flag_delivery_before_order"] = (
    clean_df["order_date"].notna()
    & (clean_df["delivery_date"] < clean_df["order_date"])
)

# A shipment can have more than one chronology issue
chronology_flags = [
    "flag_dispatch_before_order",
    "flag_delivery_before_dispatch",
    "flag_delivery_before_order"
]

clean_df["flag_any_chronology_issue"] = clean_df[
    chronology_flags
].any(axis=1)

print("DATA-QUALITY FLAG SUMMARY")
print("-" * 45)

for col in [
    "flag_missing_order_date",
    *chronology_flags,
    "flag_any_chronology_issue"
]:
    print(f"{col}: {clean_df[col].sum()}")

print(
    "\nUnique shipments with at least one chronology issue:",
    clean_df["flag_any_chronology_issue"].sum()
)

print("\nSample flagged shipments:")
display(
    clean_df.loc[
        clean_df["flag_any_chronology_issue"],
        [
            "shipment_id", "order_date", "dispatch_date",
            "delivery_date", *chronology_flags
        ]
    ].head(10)
)
```

```python
# STEP 5: CREATE CORE BUSINESS METRICS

import numpy as np
import pandas as pd

# Work on a separate copy so the cleaned dataset remains intact
analysis_df = clean_df.copy()

# 1. Total recorded shipment cost
cost_columns = [
    "fuel_cost_inr",
    "driver_cost_inr",
    "toll_cost_inr",
    "handling_cost_inr"
]

# Require all four cost components to calculate a complete total
analysis_df["recorded_total_cost_inr"] = (
    analysis_df[cost_columns].sum(axis=1, min_count=len(cost_columns))
)

# 2. Contribution and contribution margin
analysis_df["contribution_inr"] = (
    analysis_df["freight_revenue_inr"]
    - analysis_df["recorded_total_cost_inr"]
)

analysis_df["contribution_margin_pct"] = (
    analysis_df["contribution_inr"]
    / analysis_df["freight_revenue_inr"].replace(0, np.nan)
    * 100
)

# 3. Cost per kilometre
valid_distance = analysis_df["distance_km"] > 0

analysis_df["recorded_cost_per_km_inr"] = np.where(
    valid_distance,
    analysis_df["recorded_total_cost_inr"] / analysis_df["distance_km"],
    np.nan
)

# 4. Actual delivery duration
analysis_df["order_to_delivery_days"] = (
    analysis_df["delivery_date"] - analysis_df["order_date"]
).dt.total_seconds() / 86400

# 5. Delivery delay relative to promised days
analysis_df["delivery_delay_days"] = (
    analysis_df["order_to_delivery_days"] - analysis_df["promised_days"]
)

analysis_df["late_vs_promise"] = (
    analysis_df["delivery_delay_days"] > 0
).where(analysis_df["order_to_delivery_days"].notna(), pd.NA)

# 6. Vehicle capacity utilisation
analysis_df["capacity_utilisation_pct"] = (
    analysis_df["weight_kg"]
    / analysis_df["vehicle_capacity_kg"].replace(0, np.nan)
    * 100
)

# 7. Summary report
print("=" * 65)
print("CORE BUSINESS METRICS — DATA COVERAGE")
print("=" * 65)

print(f"Total shipments: {len(analysis_df)}")
print(
    "Shipments with complete cost data:",
    analysis_df["recorded_total_cost_inr"].notna().sum()
)
print(
    "Shipments with cost and positive distance:",
    analysis_df["recorded_cost_per_km_inr"].notna().sum()
)
print(
    "Shipments with valid order-to-delivery duration:",
    analysis_df["order_to_delivery_days"].notna().sum()
)
print(
    "Shipments with calculable capacity utilisation:",
    analysis_df["capacity_utilisation_pct"].notna().sum()
)

print("\n" + "=" * 65)
print("DESCRIPTIVE STATISTICS")
print("=" * 65)

metric_columns = [
    "freight_revenue_inr",
    "recorded_total_cost_inr",
    "contribution_inr",
    "contribution_margin_pct",
    "recorded_cost_per_km_inr",
    "order_to_delivery_days",
    "delivery_delay_days",
    "capacity_utilisation_pct"
]

display(
    analysis_df[metric_columns]
    .describe()
    .round(2)
)

print("\n" + "=" * 65)
print("MISSING VALUES IN DERIVED METRICS")
print("=" * 65)

print(
    analysis_df[metric_columns]
    .isna()
    .sum()
    .to_string()
)

print("\n" + "=" * 65)
print("DELIVERY STATUS COUNTS")
print("=" * 65)

print(analysis_df["delivery_status"].value_counts(dropna=False).to_string())

print("\n" + "=" * 65)
print("LATE VS PROMISE — COUNTS")
print("=" * 65)

print(
    analysis_df["late_vs_promise"]
    .value_counts(dropna=False)
    .to_string()
)
```

```python
# STEP 6: CUSTOMER SEGMENT PROFITABILITY

# Use shipments with complete financial cost data
financial_df = analysis_df[
    analysis_df["recorded_total_cost_inr"].notna()
].copy()

# Aggregate by customer segment
segment_analysis = (
    financial_df.groupby("customer_segment")
    .agg(
        shipment_count=("shipment_id", "count"),
        total_revenue=("freight_revenue_inr", "sum"),
        total_cost=("recorded_total_cost_inr", "sum"),
        total_contribution=("contribution_inr", "sum"),
        avg_revenue=("freight_revenue_inr", "mean"),
        avg_cost=("recorded_total_cost_inr", "mean"),
        avg_contribution=("contribution_inr", "mean"),
        avg_margin=("contribution_margin_pct", "mean"),
        loss_making_shipments=(
            "contribution_inr",
            lambda x: (x < 0).sum()
        )
    )
    .reset_index()
)

# Calculate contribution margin using aggregate revenue and cost
segment_analysis["aggregate_margin_pct"] = (
    segment_analysis["total_contribution"]
    / segment_analysis["total_revenue"] * 100
)

segment_analysis["loss_making_pct"] = (
    segment_analysis["loss_making_shipments"]
    / segment_analysis["shipment_count"] * 100
)

# Sort by total contribution
segment_analysis = segment_analysis.sort_values(
    "total_contribution", ascending=False
)

print("=" * 70)
print("CUSTOMER SEGMENT PROFITABILITY ANALYSIS")
print("=" * 70)

display(segment_analysis.round(2))

print("\n" + "=" * 70)
print("REVENUE CONTRIBUTION BY CUSTOMER SEGMENT")
print("=" * 70)

segment_analysis["revenue_share_pct"] = (
    segment_analysis["total_revenue"]
    / segment_analysis["total_revenue"].sum() * 100
)

display(
    segment_analysis[
        [
            "customer_segment",
            "total_revenue",
            "revenue_share_pct",
            "total_contribution",
            "aggregate_margin_pct"
        ]
    ].round(2)
)
```

```python
# STEP 10A: ROUTE-CARRIER ANALYSIS USING NORMAL MARGIN

route_carrier_df = financial_df[
    financial_df["carrier_type"].notna()
].copy()

route_carrier_analysis = (
    route_carrier_df.groupby(
        ["origin_hub", "destination_city", "carrier_type"]
    )
    .agg(
        shipment_count=("shipment_id", "count"),
        avg_revenue=("freight_revenue_inr", "mean"),
        avg_cost=("recorded_total_cost_inr", "mean"),
        avg_contribution=("contribution_inr", "mean"),
        normal_margin_pct=("contribution_margin_pct", "mean"),
        total_contribution=("contribution_inr", "sum"),
        loss_making_shipments=(
            "contribution_inr",
            lambda x: (x < 0).sum()
        ),
        avg_distance_km=("distance_km", "mean")
    )
    .reset_index()
)

route_carrier_analysis["loss_making_pct"] = (
    route_carrier_analysis["loss_making_shipments"]
    / route_carrier_analysis["shipment_count"] * 100
)

# Keep route-carrier combinations with at least 10 shipments
route_carrier_analysis = route_carrier_analysis[
    route_carrier_analysis["shipment_count"] >= 10
]

# Compare only routes where BOTH carriers have at least 10 shipments
comparison = route_carrier_analysis.pivot(
    index=["origin_hub", "destination_city"],
    columns="carrier_type",
    values=[
        "shipment_count",
        "normal_margin_pct",
        "avg_contribution",
        "avg_distance_km",
        "loss_making_pct"
    ]
)

comparison = comparison.dropna(
    subset=[
        ("normal_margin_pct", "Own Fleet"),
        ("normal_margin_pct", "Third-Party")
    ]
)

comparison["margin_difference_pp"] = (
    comparison[("normal_margin_pct", "Own Fleet")]
    - comparison[("normal_margin_pct", "Third-Party")]
)

comparison = comparison.sort_values(
    "margin_difference_pp",
    ascending=False
)

display(comparison.round(2))

print("Routes with both carriers:", len(comparison))
```

```python
# STEP 11: ROUTE-WISE CARRIER DECISION TABLE

decision_df = comparison.copy()

decision_df.columns = [
    f"{metric}_{carrier}".replace(" ", "_").lower()
    if carrier else metric
    for metric, carrier in decision_df.columns
]

decision_df = decision_df.reset_index()

decision_df["higher_margin_carrier"] = np.where(
    decision_df["margin_difference_pp"] > 0,
    "Own Fleet",
    np.where(
        decision_df["margin_difference_pp"] < 0,
        "Third-Party",
        "Equal"
    )
)

decision_df["absolute_margin_difference_pp"] = (
    decision_df["margin_difference_pp"].abs()
)

decision_df = decision_df.sort_values(
    "absolute_margin_difference_pp",
    ascending=False
)

display(decision_df.round(2))

print("\nCARRIER ADVANTAGE SUMMARY")
display(
    decision_df["higher_margin_carrier"]
    .value_counts()
    .to_frame("number_of_routes")
)
```

```python
# STEP 14: MATCHED SHIPMENT-MIX COMPARISON

matched_df = (
    mix_df.groupby(
        [
            "origin_hub",
            "destination_city",
            "commodity",
            "customer_segment",
            "vehicle_type",
            "carrier_type"
        ]
    )
    .agg(
        shipment_count=("shipment_id", "count"),
        normal_margin_pct=("contribution_margin_pct", "mean"),
        avg_contribution=("contribution_inr", "mean"),
        avg_distance_km=("distance_km", "mean"),
        avg_weight_kg=("weight_kg", "mean"),
        loss_making_pct=(
            "contribution_inr",
            lambda x: (x < 0).mean() * 100
        )
    )
    .reset_index()
)

# Keep groups with at least 5 shipments per carrier
matched_df = matched_df[
    matched_df["shipment_count"] >= 5
]

matched_comparison = matched_df.pivot(
    index=[
        "origin_hub",
        "destination_city",
        "commodity",
        "customer_segment",
        "vehicle_type"
    ],
    columns="carrier_type",
    values=[
        "shipment_count",
        "normal_margin_pct",
        "avg_contribution",
        "avg_distance_km",
        "avg_weight_kg",
        "loss_making_pct"
    ]
)

# Retain only groups where both carriers meet the threshold
matched_comparison = matched_comparison.dropna(
    subset=[
        ("normal_margin_pct", "Own Fleet"),
        ("normal_margin_pct", "Third-Party")
    ]
)

matched_comparison["margin_difference_pp"] = (
    matched_comparison[
        ("normal_margin_pct", "Own Fleet")
    ]
    - matched_comparison[
        ("normal_margin_pct", "Third-Party")
    ]
)

display(matched_comparison.round(2))

print("Matched shipment groups:", len(matched_comparison))
```

```python
# STEP 16: PREPARE REGRESSION DATA

import statsmodels.formula.api as smf

reg_df = mix_df.copy()

# Remove extreme or invalid margin observations for initial diagnostics
print("Margin distribution:")
display(
    reg_df["contribution_margin_pct"]
    .describe(percentiles=[0.01, 0.05, 0.5, 0.95, 0.99])
    .round(2)
)

print("\nNumber of margins above 100%:")
print((reg_df["contribution_margin_pct"] > 100).sum())

print("\nNumber of margins below -100%:")
print((reg_df["contribution_margin_pct"] < -100).sum())

print("\nRegression sample size:", len(reg_df))
```

```python
# STEP 17: MULTIVARIABLE OLS REGRESSION

import statsmodels.formula.api as smf

reg_df = mix_df.copy()

model = smf.ols(
    formula="""
    contribution_margin_pct ~
    C(carrier_type)
    + C(origin_hub)
    + C(destination_city)
    + C(commodity)
    + C(customer_segment)
    + C(vehicle_type)
    + distance_km
    + weight_kg
    """,
    data=reg_df
).fit(cov_type="HC3")

print(model.summary())
```

```python
# STEP 18: ROUTE FIXED-EFFECTS REGRESSION

reg_df = mix_df.copy()

# Create exact route identifier
reg_df["route"] = (
    reg_df["origin_hub"].astype(str)
    + " -> "
    + reg_df["destination_city"].astype(str)
)

# Standardise continuous variables to improve numerical conditioning
reg_df["distance_z"] = (
    reg_df["distance_km"] - reg_df["distance_km"].mean()
) / reg_df["distance_km"].std()

reg_df["weight_z"] = (
    reg_df["weight_kg"] - reg_df["weight_kg"].mean()
) / reg_df["weight_kg"].std()

# Route fixed-effects model
route_model = smf.ols(
    formula="""
    contribution_margin_pct ~
    C(carrier_type)
    + C(route)
    + C(commodity)
    + C(customer_segment)
    + C(vehicle_type)
    + distance_z
    + weight_z
    """,
    data=reg_df
).fit(
    cov_type="cluster",
    cov_kwds={
        "groups": reg_df["route"]
    }
)

print(route_model.summary())

print("\nCARRIER COEFFICIENT")
print(route_model.params["C(carrier_type)[T.Third-Party]"])

print("\n95% CONFIDENCE INTERVAL")
print(route_model.conf_int().loc[
    "C(carrier_type)[T.Third-Party]"
])
```

```python
# STEP 19: ORIGIN HUB PERFORMANCE ANALYSIS

hub_df = financial_df.copy()

hub_analysis = (
    hub_df.groupby("origin_hub")
    .agg(
        shipment_count=("shipment_id", "count"),
        total_revenue=("freight_revenue_inr", "sum"),
        total_cost=("recorded_total_cost_inr", "sum"),
        total_contribution=("contribution_inr", "sum"),
        avg_contribution=("contribution_inr", "mean"),

        # Arithmetic mean of individual shipment margins
        normal_margin_pct=("contribution_margin_pct", "mean"),

        loss_making_shipments=(
            "contribution_inr",
            lambda x: (x < 0).sum()
        ),

        avg_delivery_days=("valid_delivery_days", "mean"),

        late_shipments=(
            "valid_late_vs_promise",
            lambda x: (x == True).sum()
        ),

        eligible_delivery_records=(
            "valid_late_vs_promise",
            "count"
        )
    )
    .reset_index()
)

hub_analysis["loss_making_pct"] = (
    hub_analysis["loss_making_shipments"]
    / hub_analysis["shipment_count"] * 100
)

hub_analysis["late_delivery_pct"] = (
    hub_analysis["late_shipments"]
    / hub_analysis["eligible_delivery_records"] * 100
)

hub_analysis = hub_analysis.sort_values(
    "normal_margin_pct",
    ascending=False
)

display(hub_analysis.round(2))

print("\nTOTAL HUBS:", len(hub_analysis))
print("TOTAL SHIPMENTS:", hub_analysis["shipment_count"].sum())
```

```python
#Step 25 — Testing the relationship between utilisation and profitability
import statsmodels.formula.api as smf

util_df = analysis_df[
    analysis_df["recorded_total_cost_inr"].notna()
    & analysis_df["valid_weight"]
    & analysis_df["valid_distance"]
].copy()

util_df["route"] = (
    util_df["origin_hub"].astype(str)
    + " -> "
    + util_df["destination_city"].astype(str)
)

util_df["utilisation_z"] = (
    util_df["valid_capacity_utilisation_pct"]
    - util_df["valid_capacity_utilisation_pct"].mean()
) / util_df["valid_capacity_utilisation_pct"].std()

util_model = smf.ols(
    formula="""
    contribution_margin_pct ~
    utilisation_z
    + C(customer_segment)
    + C(route)
    + C(commodity)
    + C(vehicle_type)
    """,
    data=util_df
).fit(
    cov_type="cluster",
    cov_kwds={"groups": util_df["route"]}
)

print(util_model.summary())
```
